# 02. Data Cleaning and Order-Level Integration
This notebook cleans `geolocation` (median coordinates per zip prefix), standardizes product categories/dimensions, aggregates `order_items`, `order_payments`, and `order_reviews` to strictly one row per `order_id`, and validates zero row multiplication.

```python
import os
from pathlib import Path
if Path.cwd().name == 'notebooks':
    os.chdir(Path.cwd().parent)
import pandas as pd

master = pd.read_csv("data/processed/olist_orders_master.csv")
delivered = pd.read_csv("data/processed/olist_delivered_orders.csv")
valid_orders = master[master["item_count"].notna()]

print(f"Master dataset shape: {master.shape}")
print(f"Unique order_id count: {master['order_id'].nunique():,}")
print(f"Orders with line items: {len(valid_orders):,}")
print(f"Delivered modeling orders: {len(delivered):,}")
print(f"Total Item Revenue: ${valid_orders['total_price'].sum():,.2f}")
print(f"Total GMV (Price + Freight): ${(valid_orders['total_price'] + valid_orders['total_freight']).sum():,.2f}")
print(f"Average Order Value (AOV): ${valid_orders['total_price'].mean():,.2f}")
print(f"Average Actual Delivery Time: {delivered['actual_delivery_days'].mean():.2f} days")
print(f"Delivered Late Delivery Rate: {delivered['is_late'].mean()*100:.2f}% ({int(delivered['is_late'].sum()):,} / {len(delivered):,})")
```
