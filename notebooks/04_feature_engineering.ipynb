# 04. Leakage-Free Feature Engineering
We define the prediction point strictly at **Order Approval (`order_approved_at`)**, prior to carrier pickup and customer delivery. Post-delivery columns (`actual_delivery_days`, `delivery_delay_days`, `review_score`) are blocked via runtime assertions.

```python
import os
from pathlib import Path
if Path.cwd().name == 'notebooks':
    os.chdir(Path.cwd().parent)
import sys
from pathlib import Path
import pandas as pd

sys.path.insert(0, str(Path.cwd()))
from src.features.build_features import (
    ALL_FEATURE_COLUMNS,
    FORBIDDEN_LEAKAGE_COLUMNS,
    get_feature_matrix_and_target,
)

featured = pd.read_csv("data/processed/olist_ml_features.csv")
X, y = get_feature_matrix_and_target(featured)

print(f"Feature Matrix X shape: {X.shape}")
print(f"Target y late rate: {y.mean()*100:.2f}%")
print("Zero overlap with forbidden post-delivery columns:", len(set(X.columns).intersection(FORBIDDEN_LEAKAGE_COLUMNS)) == 0)
print("\nEngineered numerical feature summary:")
print(X.describe().T[["mean", "std", "50%", "max"]].round(2).to_string())
```
