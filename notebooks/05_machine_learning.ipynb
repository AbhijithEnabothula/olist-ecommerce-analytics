# 05. Machine Learning Model Development (Chronological Validation)
Orders are split chronologically by `order_purchase_timestamp` (oldest 80% train: `2016-09-15` to `2018-05-26`; newest 20% test: `2018-05-26` to `2018-08-29`) and evaluated with 3-fold `TimeSeriesSplit` cross-validation.

```python
import os
from pathlib import Path
if Path.cwd().name == 'notebooks':
    os.chdir(Path.cwd().parent)
import json
import pandas as pd

results = pd.read_csv("reports/model_results/model_comparison.csv")
print("=== Chronological Model Comparison Results ===")
print(results.to_string(index=False))

with open("models/saved_models/model_metadata.json") as f:
    meta = json.load(f)
print(f"\nSelected Best Model: {meta['best_model_name']}")
print(f"Train Window: {meta['train_start']} -> {meta['train_end']} ({meta['train_rows']:,} orders)")
print(f"Test Window : {meta['test_start']} -> {meta['test_end']} ({meta['test_rows']:,} orders, late rate={meta['test_late_rate']*100:.2f}%)")
```
